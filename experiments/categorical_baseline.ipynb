```python
def apply_ordered_rules(x, ordered_names, rules):

    x_new = x.copy()
    assert x_new.shape[1] == len(ordered_names)
    for j, name in enumerate(ordered_names):
        if name not in rules:
            continue

        rule = rules[name]
        for old_code, new_code in rule["remap_codes"].items():
            x_new[x_new[:, j] == old_code, j] = new_code

        for code in rule["zero_codes"]:
            x_new[x_new[:, j] == code, j] = 0

        for code in rule["nan_codes"]:
            x_new[x_new[:, j] == code, j] = np.nan

    return x_new
```

```python
def apply_decode_rules(x, decode_names, rules):
    x_new = x.copy()

    for j, name in enumerate(decode_names):
        col = x[:, j]
        rule = rules[name]

        decoded = np.full(col.shape, np.nan)
        # zero
        for code in rule["zero_codes"]:
            mask = col == code
            decoded[mask] = 0
        # remap
        for old,new in rule["remap_codes"].items():
            mask = col == old
            decoded[mask]=new

        # ranges
        for r in rule["ranges"]:
            mask = (col>=r["min"])&(col<=r["max"])
            decoded[mask]=(col[mask]-r["subtract"])*r["scale"]

        x_new[:, j] = decoded

    return x_new
```

```python
def sigmoid(t):
    t = np.clip(t, -500, 500)
    return 1 / (1 + np.exp(-t))


def calculate_logistic_loss(y, tx, w):
    """
    y:  (N,) with values {0, 1}
    tx: (N, D)
    w:  (D,)
    """
    z = tx @ w

    loss = np.mean(np.logaddexp(0, z) - y * z)

    return float(loss)


def calculate_logistic_gradient(y, tx, w):
    """
    Gradient of logistic loss.
    """
    p = sigmoid(tx @ w)

    grad = tx.T @ (p - y) / len(y)

    return grad
```

```python
def l1reg_logistic_regression(y, tx, lambda_, initial_w, max_iters, gamma):
    """
    Logistic regression using gradient descent.

    Returns:
        w, loss
    """
    w = initial_w.copy()

    for _ in range(max_iters):
        grad = calculate_logistic_gradient(y, tx, w) + lambda_*np.sign(w)
        w = w - gamma * grad

    loss = calculate_logistic_loss(y, tx, w)

    return w, loss
```

```python
def reg_logistic_regression(y, tx, lambda_, initial_w, max_iters, gamma):
    """
    Logistic regression using gradient descent.

    Returns:
        w, loss
    """
    w = initial_w.copy()

    for _ in range(max_iters):
        grad = calculate_logistic_gradient(y, tx, w) + 2*lambda_*w
        w = w - gamma * grad

    loss = calculate_logistic_loss(y, tx, w)

    return w, loss
```
