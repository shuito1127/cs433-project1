```python
def apply_ordered_rules(x, ordered_names, rules):

    x_new = x.copy()
    assert x_new.shape[1] == len(ordered_names)
    for j, name in enumerate(ordered_names):
        if name not in rules:
            continue

        rule = rules[name]
        for old_code, new_code in rule["remap_codes"].items():
            x_new[x_new[:, j] == old_code, j] = new_code

        for code in rule["zero_codes"]:
            x_new[x_new[:, j] == code, j] = 0

        for code in rule["nan_codes"]:
            x_new[x_new[:, j] == code, j] = np.nan

    return x_new
```

```python
x_ord_train = apply_ordered_rules(
    x_ord_train,
    ordered_names,
    ordered_rules
)

x_ord_val = apply_ordered_rules(
    x_ord_val,
    ordered_names,
    ordered_rules
)
```

```python
def fit_mean_imputation(x):
    mean=np.nanmean(x,axis=0)
    x_new=np.where(np.isnan(x),mean,x)
    return x_new, mean

def transform_mean_imputation(x,mean):
    x_new=np.where(np.isnan(x),mean,x)
    return x_new
```

```python
x_ord_train, ord_impute_mean = fit_mean_imputation(x_ord_train)

x_ord_val = transform_mean_imputation(
    x_ord_val,
    ord_impute_mean
)
```

```python
def fit_standardize(x):
    centered_data = np.mean(x, axis=0)
    std=np.std(x,axis=0)
    std_data = (x-centered_data) / std
    return std_data, centered_data, std

def transform_standardize(x,centered_data,std):
    return (x-centered_data)/std
```

```python
def sigmoid(t):
    # overflow対策
    t = np.clip(t, -500, 500)
    return 1 / (1 + np.exp(-t))


def calculate_logistic_loss(y, tx, w):
    """
    y:  (N,) with values {0, 1}
    tx: (N, D)
    w:  (D,)
    """
    z = tx @ w

    # BCEと同値だが、log(0)を避けるため数値的に安定
    loss = np.mean(np.logaddexp(0, z) - y * z)

    return float(loss)


def calculate_logistic_gradient(y, tx, w):
    """
    Gradient of logistic loss.
    """
    p = sigmoid(tx @ w)

    grad = tx.T @ (p - y) / len(y)

    return grad


def logistic_regression(y, tx, initial_w, max_iters, gamma):
    """
    Logistic regression using gradient descent.

    Returns:
        w, loss
    """
    w = initial_w.copy()

    for _ in range(max_iters):
        grad = calculate_logistic_gradient(y, tx, w)
        w = w - gamma * grad

    loss = calculate_logistic_loss(y, tx, w)

    return w, loss


# ==========================================
# Train on full training data
# ==========================================

# {-1, 1} -> {0, 1}
y_train_full_01 = (y_train_full == 1).astype(float)

initial_w = np.zeros(tx_train_submit.shape[1])

w_log, loss_log = logistic_regression(
    y_train_full_01,
    tx_train_submit,
    initial_w,
    max_iters=200,
    gamma=0.1
)

print("final loss:", loss_log)


# ==========================================
# Predict test data
# ==========================================

prob_test = sigmoid(tx_test_submit @ w_log)

# 0.5 is only an initial threshold
y_pred_test_log = np.where(
    prob_test >= 0.5,
    1,
    -1
)

print(np.unique(y_pred_test_log, return_counts=True))
```
