```python
def apply_ordered_rules(x, ordered_names, rules):
    x_new = x.copy()

    for j, name in enumerate(ordered_names):
        if name not in rules:
            continue

        rule = rules[name]

        for code in rule["zero_codes"]:
            x_new[x_new[:, j] == code, j] = 0

        for code in rule["nan_codes"]:
            x_new[x_new[:, j] == code, j] = np.nan

    return x_new
```

```python
x_ord_train = apply_ordered_rules(
    x_ord_train,
    ordered_names,
    ordered_rules
)

x_ord_val = apply_ordered_rules(
    x_ord_val,
    ordered_names,
    ordered_rules
)
```

```python
def fit_mean_imputation(x):
    mean=np.nanmean(x,axis=0)
    x_new=np.where(np.isnan(x),mean,x)
    return x_new, mean

def transform_mean_imputation(x,mean):
    x_new=np.where(np.isnan(x),mean,x)
    return x_new
```

```python
x_ord_train, ord_impute_mean = fit_mean_imputation(x_ord_train)

x_ord_val = transform_mean_imputation(
    x_ord_val,
    ord_impute_mean
)
```

```python
def fit_standardize(x):
    centered_data = np.mean(x, axis=0)
    std=np.std(x,axis=0)
    std_data = (x-centered_data) / std
    return std_data, centered_data, std

def transform_standardize(x,centered_data,std):
    return (x-centered_data)/std
```

```python
x_ord_train, ord_mean, ord_std = fit_standardize(x_ord_train)

x_ord_val = transform_standardize(
    x_ord_val,
    ord_mean,
    ord_std
)
```
